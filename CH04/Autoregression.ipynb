## Autoregression

- **Autoregression (AR)** is a time series modeling technique where the current value of the series is expressed as a linear combination of its past values (lags). 
- The term **"auto"** refers to the fact that the regression is performed on the same variable (the time series) using its own past values.
    - **"Auto"**: The model uses the same variable (the time series itself) as predictors. It does not rely on external variables.
- **"Regression"**: The model performs a regression analysis, where the dependent variable (( X_t )) is predicted based on its own past values .


### Key Characteristics of Autoregression
- **Lagged Dependence**: The model assumes that the current value depends on its past values.
- **Order of the Model**: The number of lagged terms used in the model is called the **"order"** (( p )).
- **Stationarity**: AR models assume that the time series is stationary (**constant mean and variance over time**). If the series is non-stationary, transformations like differencing are applied.

### Why Use Autoregression?

- **Captures Temporal Dependence**: AR models are effective for time series with strong autocorrelation.
- **Simple and Interpretable**: The coefficients directly show the influence of past values on the current value.
- **Foundation for Advanced Models**: AR is a key component of more complex models like ARIMA and SARIMA.

- In summary, we call it autoregression because it performs regression using the time series' own past values as predictors.

## Lets Simulate a AR series

```python
from statsmodels.tsa.statespace.sarimax import SARIMAX
from statsmodels.tsa.arima_process import ArmaProcess
from statsmodels.graphics.tsaplots import plot_acf
from statsmodels.tsa.stattools import adfuller
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

import warnings
warnings.filterwarnings('ignore')

%matplotlib inline
```

```python
def generate_ar_process(ar ):
    mr = np.array([1, 0, 0]) # no MA component
    MA2_process = ArmaProcess(ar, mr).generate_sample(nsample=200)
    fig, ax = plt.subplots()

    ax.plot(MA2_process)
    ax.set_xlabel('Timesteps')
    ax.set_ylabel('Value')

    fig.autofmt_xdate()
    plt.tight_layout()
```

```python
np.random.seed(42)

generate_ar_process(np.array([1, 0, 0]))
generate_ar_process(np.array([1, 0.9, 0,0]))
generate_ar_process(np.array([1, 0.9, 0,0.7]))
```
